```python
# ===========================================
# Install Required Packages
# ===========================================
!pip install --quiet tensorflow opencv-python-headless scipy matplotlib

# ===========================================
# Imports
# ===========================================
import cv2
import numpy as np
import tensorflow as tf
from google.colab import drive, files
from scipy.signal import butter, filtfilt, find_peaks
import os

# ===========================================
# Mount Google Drive
# ===========================================
drive.mount('/content/drive')

# ===========================================
# Load Model
# ===========================================
MODEL_PATH = "/content/drive/MyDrive/fake_real_model.h5"
print("Loading model...")
model = tf.keras.models.load_model(MODEL_PATH)
print("Model Loaded ✔")
class_names = ["Fake", "Real"]
IMG_SIZE = 128

# ===========================================
# Function: Upload File
# ===========================================
def upload_file():
    uploaded = files.upload()
    for filename in uploaded.keys():
        print(f"Uploaded file: {filename}")
        return filename
    return None

# ===========================================
# Function: Predict Image
# ===========================================
def predict_image(image):
    img = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    img = img.astype('float32') / 255.0
    img = np.expand_dims(img, axis=0)
    pred = model.predict(img, verbose=0)
    score = float(pred[0][0])
    label = "Real" if score > 0.5 else "Fake"
    return label, score

# ===========================================
# Function: Extract PPG from Frames
# ===========================================
def extract_ppg_from_frames(frames):
    green_signal = []
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
    for frame in frames:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.3, 5)
        if len(faces) > 0:
            x, y, w, h = faces[0]
            face_roi = frame[y:y+h, x:x+w]
            green_avg = np.mean(face_roi[:, :, 1])
            green_signal.append(green_avg)
    green_signal = np.array(green_signal)
    if len(green_signal) < 30:
        return green_signal
    fs = 30
    b, a = butter(3, [0.7/(fs/2), 4/(fs/2)], btype='band')
    filtered_signal = filtfilt(b, a, green_signal)
    return filtered_signal

# ===========================================
# Function: Compute Heart Rate
# ===========================================
def compute_ppg_features(ppg_signal, fps=30):
    if len(ppg_signal) < 2:
        return 0
    peaks, _ = find_peaks(ppg_signal, distance=fps*0.5)
    if len(peaks) < 2:
        return 0
    rr_intervals = np.diff(peaks) / fps
    heart_rate = 60 / np.mean(rr_intervals)
    return heart_rate

# ===========================================
# Function: Predict From Video
# ===========================================
def predict_from_video(video_path):
    cap = cv2.VideoCapture(video_path)
    frames = []
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frames.append(frame)
    cap.release()
    if len(frames) == 0:
        print("No frames found in video!")
        return None, None, None

    # Predict using the first frame
    label_img, score_img = predict_image(frames[0])

    # Extract PPG from multiple frames
    ppg_signal = extract_ppg_from_frames(frames)
    hr = compute_ppg_features(ppg_signal)

    final_label = label_img
    return final_label, score_img, hr

# ===========================================
# Function: Predict From Image
# ===========================================
def predict_from_image(image_path):
    image = cv2.imread(image_path)
    label, score = predict_image(image)
    return label, score, "Unavailable"  # HR not available for single image

# ===========================================
# Run Full Pipeline
# ===========================================
print("📁 Upload an image or video file for deepfake detection…")
file_path = upload_file()

if file_path.lower().endswith(('.mp4', '.avi', '.mov')):
    # Video file
    final_label, conf, hr = predict_from_video(file_path)
else:
    # Image file
    final_label, conf, hr = predict_from_image(file_path)

print("\n===== RESULT =====")
print("Prediction:", final_label)
print("Confidence:", conf)
print("Estimated Heart Rate (BPM):", hr)
print("==================")
```

```python
# ===========================================
# Install Required Packages
# ===========================================
!pip install --quiet tensorflow opencv-python-headless scipy matplotlib

# ===========================================
# Imports
# ===========================================
import cv2
import numpy as np
import tensorflow as tf
from google.colab import drive, files
from google.colab.output import eval_js
import base64
from scipy.signal import butter, filtfilt, find_peaks
import matplotlib.pyplot as plt

# ===========================================
# Mount Google Drive
# ===========================================
drive.mount('/content/drive')

# ===========================================
# Load Model
# ===========================================
MODEL_PATH = "/content/drive/MyDrive/fake_real_model.h5"
model = tf.keras.models.load_model(MODEL_PATH)
IMG_SIZE = 128

# ===========================================
# Webcam Capture (8 sec)
# ===========================================
def capture_video(duration=8, fps=30):
    js = """
    async function recordVideo(duration) {
      const video = document.createElement('video');
      const stream = await navigator.mediaDevices.getUserMedia({video: true});
      document.body.appendChild(video);
      video.srcObject = stream;
      await video.play();

      const frames = [];
      const canvas = document.createElement('canvas');
      canvas.width = video.videoWidth;
      canvas.height = video.videoHeight;
      const ctx = canvas.getContext('2d');

      const start = performance.now();
      while ((performance.now() - start) < duration * 1000) {
        ctx.drawImage(video, 0, 0, canvas.width, canvas.height);
        frames.push(canvas.toDataURL('image/jpeg', 0.8));
        await new Promise(r => setTimeout(r, 1000/30));
      }

      stream.getTracks().forEach(t => t.stop());
      video.remove();
      return frames;
    }
    recordVideo(8);
    """
    js_frames = eval_js(js)
    frames = []
    for f in js_frames:
        img = cv2.imdecode(
            np.frombuffer(base64.b64decode(f.split(',')[1]), np.uint8),
            cv2.IMREAD_COLOR
        )
        frames.append(img)
    return frames

# ===========================================
# CNN Prediction
# ===========================================
def predict_image(image):
    img = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    img = img.astype("float32") / 255.0
    img = np.expand_dims(img, axis=0)
    pred = model.predict(img, verbose=0)
    score = float(pred[0][0])
    label = "Real" if score > 0.5 else "Fake"
    return label, score

# ===========================================
# Extract PPG (Forehead ROI)
# ===========================================
def extract_ppg(frames):
    green = []
    face_cascade = cv2.CascadeClassifier(
        cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
    )

    for frame in frames:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, 1.3, 5)
        if len(faces) > 0:
            x, y, w, h = faces[0]
            fh = frame[
                y:y+int(h*0.25),
                x+int(w*0.3):x+int(w*0.7)
            ]
            if fh.size > 0:
                green.append(np.mean(fh[:,:,1]))

    green = np.array(green)
    if len(green) < 30:
        return green

    fs = 30
    b, a = butter(3, [0.7/(fs/2), 4/(fs/2)], btype="band")
    return filtfilt(b, a, green)

# ===========================================
# Heart Rate
# ===========================================
def compute_hr(ppg, fps=30):
    if len(ppg) < 30:
        return 0
    peaks, _ = find_peaks(ppg, distance=fps*0.4)
    if len(peaks) < 2:
        return 0
    hr = 60 / np.mean(np.diff(peaks)/fps)
    if hr < 40 or hr > 180:
        return 0
    return int(hr)

# ===========================================
# Display Result
# ===========================================
def show_result(image, label, conf, hr):
    img = image.copy()
    face_cascade = cv2.CascadeClassifier(
        cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
    )
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)

    for (x,y,w,h) in faces:
        cv2.rectangle(img,(x,y),(x+w,y+h),(255,0,0),2)

    color = (0,255,0) if label=="Real" else (0,0,255)
    cv2.putText(img, f"{label} ({conf:.2f})", (20,40),
                cv2.FONT_HERSHEY_SIMPLEX,1,color,2)

    text = f"HR: {hr} BPM" if hr>0 else "HR: Unavailable"
    cv2.putText(img, text, (20,80),
                cv2.FONT_HERSHEY_SIMPLEX,0.9,(255,255,0),2)

    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()

# ===========================================
# Plot PPG
# ===========================================
def plot_ppg(ppg, fps=30):
    if len(ppg) < 10:
        print("PPG not available")
        return
    t = np.arange(len(ppg))/fps
    plt.figure(figsize=(10,4))
    plt.plot(t, ppg)
    plt.xlabel("Time (s)")
    plt.ylabel("PPG Amplitude")
    plt.title("PPG Signal")
    plt.grid()
    plt.show()

# ===========================================
# MAIN MENU
# ===========================================
print("""
Choose Input Mode:
1 → Live Webcam
2 → Upload Image
3 → Upload Video
""")

choice = input("Enter choice (1/2/3): ")

# ===========================================
# OPTION 1: WEBCAM
# ===========================================
if choice == "1":
    frames = capture_video()
    label, conf = predict_image(frames[0])
    ppg = extract_ppg(frames)
    hr = compute_hr(ppg)

    show_result(frames[0], label, conf, hr)
    plot_ppg(ppg)

# ===========================================
# OPTION 2: IMAGE UPLOAD
# ===========================================
elif choice == "2":
    uploaded = files.upload()
    path = list(uploaded.keys())[0]
    image = cv2.imread(path)

    label, conf = predict_image(image)
    show_result(image, label, conf, hr=0)

    print("PPG not applicable for single image")

# ===========================================
# OPTION 3: VIDEO UPLOAD
# ===========================================
elif choice == "3":
    uploaded = files.upload()
    path = list(uploaded.keys())[0]

    cap = cv2.VideoCapture(path)
    frames = []
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        frames.append(frame)
    cap.release()

    label, conf = predict_image(frames[0])
    ppg = extract_ppg(frames)
    hr = compute_hr(ppg)

    show_result(frames[0], label, conf, hr)
    plot_ppg(ppg)

else:
    print("Invalid choice")
```
